```python

careers = {
    "Data Scientist": {
        "skills": ["Programming", "Mathematics"],
        "interest": "Technology",
        "description": "Analyzes data to solve problems.",
        "required": "Python, Mathematics, Statistics",
        "scope": "AI, data analytics, and business intelligence"
    },
    "AI Engineer": {
        "skills": ["Programming", "Mathematics"],
        "interest": "Technology",
        "description": "Builds applications using artificial intelligence.",
        "required": "Python, Machine Learning, Mathematics",
        "scope": "AI applications and automation"
    },
    "Web Developer": {
        "skills": ["Programming", "Problem Solving"],
        "interest": "Technology",
        "description": "Creates websites and web applications.",
        "required": "HTML, CSS, JavaScript",
        "scope": "Web development and freelancing"
    },
    "Doctor": {
        "skills": ["Communication", "Research"],
        "interest": "Healthcare",
        "description": "Diagnoses illnesses and treats patients.",
        "required": "Biology, Medical Knowledge, Communication",
        "scope": "Hospitals and healthcare"
    },
    "Graphic Designer": {
        "skills": ["Creativity", "Designing"],
        "interest": "Arts",
        "description": "Creates visual designs and graphics.",
        "required": "Creativity, Design Software, Color Theory",
        "scope": "Advertising and digital design"
    },
    "Digital Marketer": {
        "skills": ["Communication", "Creativity"],
        "interest": "Business",
        "description": "Promotes products and services online.",
        "required": "SEO, Marketing, Communication",
        "scope": "Digital marketing and advertising"
    },
    "Cybersecurity Analyst": {
        "skills": ["Problem Solving", "Networking"],
        "interest": "Technology",
        "description": "Protects systems from cyber threats.",
        "required": "Networking, Linux, Security",
        "scope": "Cybersecurity and information security"
    },
    "Software Engineer": {
        "skills": ["Programming", "Problem Solving"],
        "interest": "Technology",
        "description": "Designs and develops software applications.",
        "required": "Programming, Algorithms, Databases",
        "scope": "Software companies and technology"
    },
    "Accountant": {
        "skills": ["Mathematics", "Organization"],
        "interest": "Business",
        "description": "Manages financial records and reports.",
        "required": "Accounting, Mathematics, Excel",
        "scope": "Finance and auditing"
    },
    "Teacher": {
        "skills": ["Communication", "Teaching"],
        "interest": "Teaching",
        "description": "Teaches students and shares knowledge.",
        "required": "Subject Knowledge, Communication, Teaching",
        "scope": "Schools and educational institutions"
    }
}


def get_input():
    name = input("\nEnter your name: ")
    age = input("Enter your age: ")
    qualification = input("Enter your qualification: ")

    print("\nSubjects:")
    print("Mathematics, Computer Science, Biology, Arts, Business")

    favorite_subject = input("Favorite subject: ")
    strongest_subject = input("Strongest subject: ")

    print("\nSkills:")
    print("Programming, Mathematics, Problem Solving, Communication,")
    print("Creativity, Designing, Research, Networking, Organization, Teaching")

    skills = input("Enter your skills separated by commas: ").split(",")

    for i in range(len(skills)):
        skills[i] = skills[i].strip().title()

    print("\nInterests:")
    print("Technology, Healthcare, Business, Arts, Teaching")

    interest = input("Your interest: ").strip().title()

    return name, age, qualification, favorite_subject, strongest_subject, skills, interest


def recommend_careers(skills, interest):
    recommendations = []

    for career in careers:
        score = 0

        for skill in skills:
            if skill in careers[career]["skills"]:
                score += 1

        if interest == careers[career]["interest"]:
            score += 2

        if score > 0:
            recommendations.append([career, score])

    recommendations.sort(key=lambda x: x[1], reverse=True)

    return recommendations


def display_report(name, age, qualification, favorite_subject,
                   strongest_subject, skills, interest, recommendations):

    print("\n========== CAREER ANALYSIS REPORT ==========")
    print("Name:", name)
    print("Age:", age)
    print("Qualification:", qualification)
    print("Favorite Subject:", favorite_subject)
    print("Strongest Subject:", strongest_subject)
    print("Skills:", ", ".join(skills))
    print("Interest:", interest)

    print("\nRecommended Careers:")

    if len(recommendations) == 0:
        print("No matching careers found. Try different skills or interests.")

    for i in range(min(3, len(recommendations))):
        career = recommendations[i][0]

        print("\n", i + 1, ".", career)
        print("Match Score:", recommendations[i][1])
        print("Description:", careers[career]["description"])
        print("Required Skills:", careers[career]["required"])
        print("Future Scope:", careers[career]["scope"])

    print("\n============================================")


def main():
    while True:
        print("\n============================================")
        print("  AI-BASED CAREER RECOMMENDATION SYSTEM")
        print("============================================")

        name, age, qualification, favorite_subject, strongest_subject, skills, interest = get_input()

        recommendations = recommend_careers(skills, interest)

        display_report(
            name, age, qualification, favorite_subject,
            strongest_subject, skills, interest, recommendations
        )

        again = input("\nWould you like to try again? (yes/no): ").strip().lower()

        if again != "yes":
            print("Thank you for using the Career Recommendation System!")
            break


main()
```
