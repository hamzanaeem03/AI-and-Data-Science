```python
# Student Task 1 — Age Category
# Print 'Child' if age < 18 else 'Adult'


def age_check(age):
  if age >= 18:
    print("adult")

  else:
     print("child")




age = int(input("Enter the age: "))

age_check(age)
```

```python
# Student Task 2 — Temperature Check
# Print "It's hot." if temperature > 30 else "It's cool."



def temp_check(temp):
  if temp > 30:
    print("It's hot.")

  else:
     print("It's cool.")




temp = int(input("Enter the Temperature: "))

temp_check(temp)
```

```python
# Student Task 3 — Divisible by 5
# Print 'Divisible by 5' or 'Not divisible by 5'


def Divisible_by_5(number):
  if number % 5:
    print("Divisible by 5")

  else:
     print("ot divisible by 5")




number = int(input("Enter the number: "))

Divisible_by_5(number)
```

```python
# Student Task 4 — Positive or Negative
# Print 'Positive' if >0 else 'Negative or Zero'


def Positive_check(number):
  if number > 0:
    print("Positive")

  else:
     print("Negative or Zero")




number = float(input("Enter the number: "))

Positive_check(number)
```
