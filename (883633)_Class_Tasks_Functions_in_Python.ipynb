```python
# Task 2 — Rectangle Calculator
# Create variables for length and width. Calculate area and perimeter

length = int(input("Enter the length "))
width = int(input("Enter the width "))
area = length * width
perimeter = 2 * (length + width)
print("area", area)
print("perimeter", perimeter)
```

```python
# Task 3 — Circle Calculator
# Create a radius variable and use PI = 3.14159 to calculate the area of a circle.

PI = 3.14159
radius = int(input("Enter the radius "))
area = 2 * PI * radius * radius
print("area", area)
```

```python
# Task 5 — Temperature Conversion  Create a Celsius temperature variable and convert it into Fahrenheit.
# Formula: Fahrenheit = (Celsius × 9/5) + 32

Celsius = 47
Fahrenheit = (Celsius * 9/5) + 32
print("Fahrenheit", Fahrenheit)
```
