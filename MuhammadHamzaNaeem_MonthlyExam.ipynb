```python
names = ["Hamza", "Ali", "Ahmed", "Raza", "Bilal"]
ages = [25, 42, 67, 31, 55]
temps = [98, 100, 101, 98, 100]
bp = [115, 128, 145, 118, 135]

normal = 0
attention = 0
critical = 0
total_temp= 0

for i in range(len(names)):

    name = names[i]
    age = ages[i]
    temperature = temps[i]
    blood_pressure = bp[i]

    total_temp = total_temp + temperature

    if (temps[i] >= 97 and temps[i] <= 99.5) and bp[i] < 120:
        status = "Normal"
        normal = normal + 1

    elif ((temps[i] > 99.5 and temps[i] < 101) or
          (bp[i] >= 120 and bp[i] <= 139)):
        status = "Needs Attention"
        attention = attention + 1

    elif temps[i] >= 101 or bp[i] >= 140:
        status = "Critical"
        critical = critical + 1

    print("Name: " + names[i])
    print("Age: " + str(ages[i]))
    print("Temperature: " + str(temps[i]))
    print("Systolic Blood Pressure: " + str(bp[i]))
    print("Status: " + str(status))
    print("----------------------------")


average_temperature = total_temp / len(names)

print("Healthcare Monitoring Summary")
print("----------------------------")
print("Total Patients: " + str(len(names)))
print("Normal: " + str(normal))
print("Needs Attention: " + str(attention))
print("Critical: " + str(critical))
print("Average Body Temperature: " + str(round(average_temperature)))
```
