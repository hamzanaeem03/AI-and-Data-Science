```python
# Project 01: Student ID & Name Analyzer
# • Create a Python program that stores a student's full name and student ID in variables.
# • Convert the student's name to uppercase and title case.
# • Remove unnecessary spaces using .strip().
# • Display the first and last character of the name using indexing.
# • Extract the first 3 and last 3 characters using slicing.
# • Extract specific sections of the student ID using slicing.
# • Display all results using print().




full_name = str(input("Enter you full name: "))
student_id = str(input("Enter you student ID: "))


print("name in uppercase: ", full_name.upper())

print("name in title case: ", full_name.title())

print("name in space stripe: ", full_name.strip())

print("letter in first position: ", full_name[0],"name in last position: ", full_name[-1] )

print("letter in first 3: ", full_name[:3],"name in last 3: ", full_name[-3:] )

print("letter in student ID: ", student_id[:4],"number in student ID: ", student_id[4:] )
```

```python
# Project 02: Email & Username Analyzer
# • Create a Python program that stores a user's full name and email address in variables.
# • Remove unnecessary spaces using .strip().
# • Convert the name to lowercase and title case.
# • Display specific characters from the name using indexing.
# • Extract the username from the email address using slicing.
# • Extract the domain name using slicing.
# • Display the complete analyzed information using print().




full_name = str(input("Enter you full name: "))
email = str(input("Enter you email : "))


print("name in uppercase: ", full_name.upper())

print("name in title case: ", full_name.title())

print("name in space stripe: ", full_name.strip())

print("letter in first position: ", full_name[0],"name in last position: ", full_name[-1] )

at_position = email.index("@")
print("Username:", email[:at_position])
print("Domain:", email[at_position + 1:])
```
