## 8. Student Tasks ✍️
1. From 'DataScience', extract 'Data', 'Science', 'ataSci'.
2. From 'Programming', extract 'ing' and 'gram' using negative indexing.
3. From 'ABCDEFGHIJ', print every 2nd char, every 3rd char, and reverse.
4. Take user input and print first 3, last 3, and middle part.
5. Reverse each word in 'Python is fun' → 'nohtyP si nuf'.

```python
# 1. From 'DataScience', extract 'Data', 'Science', 'ataSci'.

text= 'DataScience'

print(text[:4])
print(text[4:])
print(text[1:7])
```

```python
# From 'Programming', extract 'ing' and 'gram' using negative indexing.
text = 'Programming'
print(text[-3:])
print(text[-8:-4])
```

```python
# 3. From 'ABCDEFGHIJ', print every 2nd char, every 3rd char, and reverse
text = 'ABCDEFGHIJ'
print(text[::2])
print(text[::3])
print(text[::-1])
```

```python
# 4. Take user input and print first 3, last 3, and middle part
text = input("Enter a string: ")

print("First 3:", text[:3])
print("Last 3:", text[-3:])

mid = len(text)
print("Middle part:", text[mid-1:mid+2])
```

```python
# 5. Reverse each word in 'Python is fun'
text = "Python is fun"

text = text.split()
text = [word[::-1] for word in text]

print(" ".join(text))
```
