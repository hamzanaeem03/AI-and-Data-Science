```python
# 8. A car travels 360 kilometers using 30 liters of fuel. Store the distance and fuel used in variables and calculate and print the kilometers traveled per liter.


km_travelled = 360
fuel_used = 30

average = km_travelled/fuel_used
print("kilometers traveled per liter", average)
```

```python
# 9. A teacher wants to divide 57 students equally into groups of 5. Store the number of students and group size in variables and use mathematical operators to print the number of complete groups and remaining students.


number_of_students = 57
number_of_groups = 5
number_of_complete_groups = number_of_students // number_of_groups
number_of_remaining_students = number_of_students % number_of_groups

print("number of complete groups", number_of_complete_groups)
print("number of remaining students", number_of_remaining_students)
```
