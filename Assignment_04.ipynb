```python
# PROJECT SPECIFICATION & REQUIREMENT DOCUMENT
# Automated Inventory Stock Monitoring System Domain: Supply Chain & Logistics
# Core Constructs: Lists, For Loops, Conditionals

# 1. EXECUTIVE SUMMARY & PROBLEM STATEMENT
# In modern retail and warehouse operations, tracking inventory levels manually across numerous Stock Keeping Units (SKUs) frequently leads to stockouts, delayed restocking orders, operational bottlenecks, and costly human errors. Operational teams require a streamlined, automated software solution to evaluate inventory counts against minimum thresholds in real time.
# The objective of this project is to develop an Automated Inventory Stock Monitor in Python. The system processes structural inventory data, classifies item statuses based on defined inventory parameters, and generates an actionable replenishment report.
# 2. FUNCTIONAL REQUIREMENTS
# ■ Data Structure & Storage
# Maintain item records using nested list structures containing product attributes: SKU/Product Name, Current Quantity, and Minimum Reorder Threshold.
# ■ Batch Processing Execution
# Iterate sequentially through the product dataset using a standard for loop to analyze each item in a single execution pass.
# ■ Status Classification Engine
# Apply conditional control flow logic (if, elif, else) to evaluate inventory levels against specified thresholds.
# ■ Alert & Summary Reporting
# Output an itemized stock summary, compile a specialized action-list for restocks, and output aggregate operational metrics.

# 3. BUSINESS RULES & LOGIC MATRIX
# STATUS CATEGORY CONDITION RULE OPERATIONAL ACTION
# OUT OF STOCK Current Quantity == 0 Immediate emergency reorder required. Flagged high priority.
# LOW STOCK 0 < Current Quantity ≤ Reorder Threshold Standard reorder placed on purchase list.
# ADEQUATE Current Quantity > Reorder Threshold No action required. Stock level optimal.
# Implementation Scope Constraint: Solution must rely strictly on Python standard constructs—utilizing lists, for loops, and conditional statements—without leveraging external libraries like Pandas or NumPy.
```

```python
inventory = [
    ["SKU101 - Laptop", 0, 10],
    ["SKU102 - Mouse", 5, 10],
    ["SKU103 - Keyboard", 25, 10],
    ["SKU104 - Monitor", 3, 5],
    ["SKU105 - Printer", 12, 8],
    ["SKU106 - Headphones", 0, 6],
    ["SKU107 - Webcam", 7, 7]
]
restock_list = []
out_of_stock_count = 0
low_stock_count = 0
adequate_stock_count = 0
total_inventory_units = 0

print("       AUTOMATED INVENTORY STOCK MONITORING SYSTEM")
for item in inventory:

    product_name = item[0]
    current_quantity = item[1]
    reorder_threshold = item[2]

    total_inventory_units += current_quantity
    if current_quantity == 0:
        status = "OUT OF STOCK"
        action = "URGENT REORDER"
        priority = "HIGH"
        out_of_stock_count += 1
        restock_list.append(
            [product_name, current_quantity,
             reorder_threshold, status, priority]
        )

    elif current_quantity <= reorder_threshold:
        status = "LOW STOCK"
        action = "REORDER REQUIRED"
        priority = "NORMAL"

        low_stock_count += 1

        restock_list.append(
            [product_name, current_quantity,
             reorder_threshold, status, priority]
        )

    else:
        status = "ADEQUATE"
        action = "NO ACTION REQUIRED"
        priority = "NONE"

        adequate_stock_count += 1
    print(f"\nProduct: {product_name}")
    print(f"Current Quantity: {current_quantity}")
    print(f"Reorder Threshold: {reorder_threshold}")
    print(f"Status: {status}")
    print(f"Action: {action}")
print("              REPLENISHMENT ACTION LIST")

if len(restock_list) == 0:
    print("No products require replenishment.")

else:
    for item in restock_list:
        print(f"\nProduct: {item[0]}")
        print(f"Available Units: {item[1]}")
        print(f"Minimum Threshold: {item[2]}")
        print(f"Status: {item[3]}")
        print(f"Priority: {item[4]}")
total_products = len(inventory)
total_restock_items = len(restock_list)

print("             OPERATIONAL SUMMARY REPORT")
print(f"Total Products Monitored: {total_products}")
print(f"Total Inventory Units: {total_inventory_units}")
print(f"Out of Stock Products: {out_of_stock_count}")
print(f"Low Stock Products: {low_stock_count}")
print(f"Adequate Stock Products: {adequate_stock_count}")
print(f"Products Requiring Reorder: {total_restock_items}")
```
