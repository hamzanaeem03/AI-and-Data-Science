```python
# Question 1 – Student Result Management System
# A school wants to develop a simple student result management system. The marks obtained by a student in five
# subjects are stored in the following list:
# marks = [78, 65, 82, 55, 90]
# Average Marks Performance
# 80 or above Excellent
# 60–79 Good
# 40–59 Pass
# Below 40 Fail
# Develop a Python program that calculates the total marks, calculates the average marks, and determines the
# student's performance according to the criteria above. Display the total marks, average marks, and performance.


marks = [78, 65, 82, 55, 90]
total_marks = sum(marks)
print("Total marks", total_marks)
average_marks = total_marks/len(marks)
print("average marks", average_marks)
if average_marks >= 80:
  print("Performance", "Excellent")
elif average_marks >= 60 and average_marks <= 79 :
  print("Performance", "Good")
elif average_marks >= 40 and average_marks <= 59 :
  print("Performance", "Pass")
elif average_marks < 40 :
  print("Performance", "Fail")
```

```python
# Question 2 – Online Shopping Discount System
# An online shopping store maintains the prices of five products in a list:
# prices = [2500, 1800, 3200, 1500, 900]
# Purchase Amount Discount
# 10,000 or above 20%
# 5,000–9,999 10%
# Below 5,000 No Discount
# Develop a Python program that calculates the total purchase amount and applies the appropriate discount. Display
# the original purchase amount, discount percentage, discount amount, and final payable amount.

prices = [2500, 1800, 3200, 1500, 900]

purchase_amount = sum(prices)

if purchase_amount >= 10000:
  discount_percentage = 20
elif purchase_amount >= 5000 and purchase_amount <= 9999 :
  discount_percentage = 10
elif purchase_amount < 5000 :
  discount_percentage = 0
discount_amount = purchase_amount/100
discount_amount = discount_amount * discount_percentage
final_payable_amount = purchase_amount - discount_amount

print("original purchase amount", purchase_amount)
print("discount percentage", discount_percentage)
print("discount amount", discount_amount)
print("final payable amount", final_payable_amount)
```

```python
# Question 3 – Cinema Seat Availability System
# A cinema maintains a list of currently available seats:
# seats = ["A1", "A2", "A3", "B1", "B2"]
# Develop a Python program that asks the customer to enter a seat number. Check whether the requested seat
# exists in the list. Display "Seat Available" if it exists; otherwise display "Seat Not Available".
# Constraint: Do not use loops.

seats = ["A1", "A2", "A3", "B1", "B2"]

user_seat = input("enter a seat number : ")

if user_seat in seats:
  print("Seat Available")
else:
  print("Seat Not Available")
```

```python
# Question 4 – Employee Performance Evaluation
# A company's HR department stores an employee's monthly performance scores in a list:
# scores = [85, 72, 90, 65, 88]
# Average Score Performance Category
# 85 or above Outstanding
# 70–84 Good
# 50–69 Needs Improvement
# Below 50 Poor
# Develop a Python program that calculates the employee's average performance score and categorizes the
# employee according to the criteria above. Display the average score and corresponding performance category.


scores = [85, 72, 90, 65, 88]
total_scores = sum(scores)
print("Total scores", total_scores)
average_scores = total_scores/len(scores)
print("average scores", average_scores)
if average_scores >= 85:
  print("Outstanding")
elif average_scores >= 70 and average_scores <= 84 :
  print("Good")
elif average_scores >= 50 and average_scores <= 69 :
  print("Needs Improvement")
elif average_scores < 50 :
  print("Poor")
```

```python
# Question 5 – Restaurant Menu and Order System
# A restaurant maintains its menu and corresponding prices using two lists:
# items = ["Burger", "Pizza", "Biryani", "Pasta"]
# prices = [500, 1200, 450, 800]
# Develop a Python program that asks the customer to enter the name of an item. Check whether the item is
# available. If available, display its price; otherwise display "Item Not Available". Use conditional statements and list
# indexing to determine the price.
# Constraint: Do not use loops.

items = ["Burger", "Pizza", "Biryani", "Pasta"]
prices = [500, 1200, 450, 800]
item_name = input("Enter item name: ")

if item_name in items:
    item_index = items.index(item_name)
    print("Price:", prices[item_index])
else:
    print("Item Not Available")
```

```python
# Question 6 – Temperature Monitoring System
# A weather monitoring application stores daily temperatures in a list:
# temperatures = [38, 41, 35, 29, 42]
# Temperature Classification
# Above 40°C Very Hot
# 30–40°C Hot
# Below 30°C Cool
# Develop a Python program that checks the first, middle, and last temperatures using list indexing. Classify each
# temperature according to the criteria above and display the temperature with its classification.
# Constraint: Do not use loops.

temperatures = [38, 41, 35, 29, 42]
first = temperatures[0]
if first > 40:
    print(first, "Very Hot")
elif first >= 30:
    print(first, "Hot")
else:
    print(first, "Cool")
middle = temperatures[2]
if middle > 40:
    print(middle, "Very Hot")
elif middle >= 30:
    print(middle, "Hot")
else:
    print(middle, "Cool")
last = temperatures[-1]
if last > 40:
    print(last, "Very Hot")
elif last >= 30:
    print(last, "Hot")
else:
    print(last, "Cool")
```

```python
# Question 7 – Bank Transaction Classification System
# A bank stores three recent transactions in a list:
# transactions = [5000, -2000, 15000]
# Transaction Amount Category
# Above 10,000 Large Transaction
# 5,000–10,000 Medium Transaction
# Below 5,000 Small Transaction
# Develop a Python program that examines the third transaction and determines its category. Also determine
# whether the transaction represents a Deposit (positive amount) or Withdrawal (negative amount). Display the
# transaction amount and its classification.

transactions = [5000, -2000, 15000]
transaction = transactions[2]
print("Transaction Amount:", transaction)
if transaction > 10000:
    print("Category: Large Transaction")
elif transaction >= 5000:
    print("Category: Medium Transaction")
else:
    print("Category: Small Transaction")
if transaction > 0:
    print("Type: Deposit")
else:
    print("Type: Withdrawal")
```

```python
# Question 8 – Student Attendance Management System
# A student's attendance record for five classes is stored in a list. P represents Present and A represents Absent:
# attendance = ["P", "P", "A", "P", "A"]
# Attendance Percentage Status
# 80% or above Eligible for Exam
# 60–79% Warning
# Below 60% Not Eligible
# Develop a Python program that calculates the number of classes attended, calculates the attendance percentage,
# and determines eligibility according to the criteria above.
# Hint: You may use the count() list method.
# Constraint: Do not use loops.


attendance = ["P", "P", "A", "P", "A"]

attended = attendance.count("P")
total_classes = len(attendance)

attendance_percentage = (attended / total_classes) * 100

if attendance_percentage >= 80:
    status = "Eligible for Exam"
elif attendance_percentage >= 60:
    status = "Warning"
else:
    status = "Not Eligible"

print("Classes Attended:", attended)
print("Attendance Percentage:", attendance_percentage, "%")
print("Status:", status)
```

```python
# Question 9 – Grocery Store Inventory System
# A grocery store maintains product names and their available stock quantities using two lists:
# products = ["Milk", "Bread", "Eggs", "Juice"]
# stock = [10, 0, 25, 3]
# Condition Status
# Product does not exist Product Not Found
# Stock = 0 Out of Stock
# Stock = 1–5 Low Stock
# Stock > 5 In Stock
# Develop a Python program that asks the user to enter a product name. Determine the product's inventory status
# according to the criteria above. Display the product name, available quantity, and inventory status where
# applicable.
# Constraint: Do not use loops.


products = ["Milk", "Bread", "Eggs", "Juice"]
stock = [10, 0, 25, 3]

product = input("Enter product name: ")

if product in products:
    index = products.index(product)
    quantity = stock[index]

    if quantity == 0:
        status = "Out of Stock"
    elif quantity <= 5:
        status = "Low Stock"
    else:
        status = "In Stock"

    print("Product:", product)
    print("Available Quantity:", quantity)
    print("Inventory Status:", status)
else:
    print("Product Not Found")
```

```python
# Question 10 – Mobile Data Usage Monitoring System

# A mobile application records a user's daily data usage for one week in GB:
# data_usage = [2.5, 4.2, 7.8, 3.1, 9.5, 5.6, 1.8]
# Weekly Usage Category
# More than 30 GB High Usage
# 20–30 GB Moderate Usage
# Below 20 GB Low Usage
# Develop a Python program that calculates the total weekly data usage, determines the usage category, and
# identifies the highest and lowest data usage values.
# Constraints: Do not use loops, sum(), max(), or min(). Display the total usage, usage category, highest usage,
# and lowest usage.



data_usage = [2.5, 4.2, 7.8, 3.1, 9.5, 5.6, 1.8]

total = (
    data_usage[0]
    + data_usage[1]
    + data_usage[2]
    + data_usage[3]
    + data_usage[4]
    + data_usage[5]
    + data_usage[6]
)

if total > 30:
    category = "High Usage"
elif total >= 20:
    category = "Moderate Usage"
else:
    category = "Low Usage"

highest = data_usage[0]

if data_usage[1] > highest:
    highest = data_usage[1]

if data_usage[2] > highest:
    highest = data_usage[2]

if data_usage[3] > highest:
    highest = data_usage[3]

if data_usage[4] > highest:
    highest = data_usage[4]

if data_usage[5] > highest:
    highest = data_usage[5]

if data_usage[6] > highest:
    highest = data_usage[6]

lowest = data_usage[0]

if data_usage[1] < lowest:
    lowest = data_usage[1]

if data_usage[2] < lowest:
    lowest = data_usage[2]

if data_usage[3] < lowest:
    lowest = data_usage[3]

if data_usage[4] < lowest:
    lowest = data_usage[4]

if data_usage[5] < lowest:
    lowest = data_usage[5]

if data_usage[6] < lowest:
    lowest = data_usage[6]

print("Total Weekly Usage:", total, "GB")
print("Usage Category:", category)
print("Highest Data Usage:", highest, "GB")
print("Lowest Data Usage:", lowest, "GB")
```
